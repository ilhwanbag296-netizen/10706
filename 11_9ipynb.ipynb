```python
from gtts import gTTS
from IPython.display import Audio
while True:
  text = input('안녕하세요 tts입니다.')
  if text =='종료':
    print('TTS 종료')
    break
  tts=gTTS(text,lang='ko')
  tts.save('output.mp3')
  display(Audio('output.mp3',autoplay=True))
```
