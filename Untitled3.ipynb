```python
def calculator(a,oper,b):
  if oper=='+':
    return a+b

  elif oper=='-':
    return a-b

  elif oper=='*':
    return a*b

  elif oper=='/':
    return a/b


  else:
    return "오류: 지원하지 않았음"



```
