```python
import pandas as pd
df = pd.read_csv('2022년 대기 정보.csv')
df['연월일']=df['연월일시'].str.split(':').str[0]
df['시']=df['연월일시'].str.split(':').str['1']
df['연월일']=pd.to_datetime(df['연월일'])
df['연도']=df['연월일'].dt.year
df['월']=df['연월일'].dt.month
df['일']=df['연월일'].dt.day
df['요일']=df['연월일'].dt.dayofweek
df.head()
def season(x):
  if 3<=x<=5:return '1_spring'
  if 6<=x<=8:return '2_spring'
  if 9<=x<=11:return '3_spring'
  else: return '4_sinter'

df['계절']= df['월'].apply(season)
df.sample(n=5)

```
